## Task 2(c): Take a grayscale image of size 512x512, find the difference image between the original & the image obtained by last three MSBs

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
dollar_image = cv2.imread("./Dollar 1192x500.tif", 0)
plt.imshow(dollar_image, cmap='gray')
plt.title("The 100 dollar Image")
plt.show()
```

## Image Obatained by last Three bit(MSB)

```python
mask = "11100000"
three_bit_image = dollar_image & int(mask, 2)
plt.imshow(three_bit_image, cmap='gray')
plt.title("Three Bit(MSB) image")
plt.show()
```

```python
difference_between_image = cv2.absdiff(np.array(dollar_image), np.array(three_bit_image))
plt.imshow(difference_between_image, cmap='gray')
plt.title("Difference between original and three bit(MSB) image")
plt.show()
```
