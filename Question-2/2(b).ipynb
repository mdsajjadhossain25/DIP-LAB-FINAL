## Task 2(b): Take a grayscale image of size 512x512, differentiate the results of power law & inverse logarithmic transformation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def power_law_transformation(image, gamma):
    transformed_image = image.copy()
    height, width = image.shape

    for r in range(height):
        for c in range(width):
            pixel = float(image[r,c])/255.0
            transformed_image[r,c] = (pixel**gamma) * 255
    return np.uint8(transformed_image)
```

```python
def inverse_log_transform(image):
    c = 255/np.log(256)
    transformed_image = np.exp(image/c) - 1

    return np.uint8(transformed_image)
```

## Apply Power Law Transform(gamma<1)

```python
spine_image = cv2.imread("./Fractured Spine 746x976.tif", 0)
gammas , subplot_number = [1, 0.6, 0.4, 0.3], 1
plt.figure(figsize=(14, 14))

for gamma in (gammas):
    transformed_image = power_law_transformation(spine_image, gamma)
    plt.subplot(2,2,subplot_number)
    plt.imshow(transformed_image, cmap='gray')
    plt.title(f"Power Law Transform with gamma = {gamma}")
    subplot_number += 1

plt.show()
```

## Applying power law transform gamma > 1

```python
aerial_image = cv2.imread("./Aerial Image 765x769 (1).tif", 0)
gammas , subplot_number = [1, 3, 4, 5], 1
plt.figure(figsize=(14, 14))

for gamma in (gammas):
    transformed_image = power_law_transformation(aerial_image, gamma)
    plt.subplot(2,2,subplot_number)
    plt.imshow(transformed_image, cmap='gray')
    plt.title(f"Power Law Transform with gamma = {gamma}")
    subplot_number += 1

plt.show()
```

## Inverse Log Transform

```python
transformed_image = inverse_log_transform(aerial_image)
plt.imshow(transformed_image, cmap='gray')
plt.title("Inverse Log Transform")
plt.show()
```
