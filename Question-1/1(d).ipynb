## Illustrate the histogram of the image and make single threshold segmentation observed from the histogram


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def generate_histogram(image):
    gray_level_count = np.zeros(256)
    height, width = image.shape

    for r in range(height):
        for c in range(width):
            gray_level_count[image[r,c]] += 1

    plt.bar(range(256), gray_level_count, width=1.0, color = 'gray')
    plt.title("The Histogram of the image")
    plt.show()
```

```python
original_image = cv2.imread("./Skeleton 750x1482.tif", 0)
plt.imshow(original_image, cmap='gray')
plt.title("The Skeleton Image")
plt.show()
```

```python
generate_histogram(original_image)
```

## Making Single Threshold Segmentation observed from the Histogram

```python
threshold_value = 27
segmented_image = np.where(original_image<threshold_value , 0, 255)
segmented_image = np.uint8(segmented_image)

plt.imshow(segmented_image, cmap='gray')
plt.title("The Segmented Image")
plt.show()
```

```python
generate_histogram(segmented_image)
```
